### 1. Memuat Dataset Hasil Cleaning

Tahap ini dilakukan untuk memuat dataset Netflix Original yang telah melalui proses Data Cleaning. Dataset hasil cleaning digunakan agar analisis dilakukan pada data yang telah diperiksa dan diperbaiki sebelumnya.

```python
import pandas as pd

# Membaca dataset hasil cleaning
df_clean = pd.read_csv(
    "../data/netflix_clean.csv",
    parse_dates=["Premiere"]
)

# Memeriksa dataset
print("Ukuran dataset:", df_clean.shape)
display(df_clean.head())
```

### 2. Analisis Bahasa Film Netflix Original

Analisis ini bertujuan untuk mengetahui bahasa yang paling banyak digunakan dalam film Netflix Original. Jumlah film dihitung berdasarkan kategori bahasa, kemudian diurutkan dari jumlah terbanyak hingga paling sedikit.

```python
# Menghitung jumlah film berdasarkan bahasa
jumlah_bahasa = (
    df_clean["Language"]
    .value_counts()
    .reset_index()
)

jumlah_bahasa.columns = ["Language", "Jumlah Film"]

# Menampilkan 10 bahasa terbanyak
display(jumlah_bahasa.head(10))

# Menampilkan bahasa yang paling banyak digunakan
print("Bahasa terbanyak:",
      jumlah_bahasa.iloc[0]["Language"])

print("Jumlah film:",
      jumlah_bahasa.iloc[0]["Jumlah Film"])
```

### 3. Analisis Film dengan Skor IMDb 8.0 ke Atas

Analisis ini dilakukan untuk mengidentifikasi film Netflix Original yang memperoleh skor IMDb minimal 8.0. Data difilter berdasarkan skor IMDb dan diurutkan dari skor tertinggi untuk mengetahui film dengan penilaian tertinggi.

```python
# Memfilter film dengan skor IMDb minimal 8.0
film_imdb_tinggi = df_clean[
    df_clean["IMDB Score"] >= 8.0
].copy()

# Mengurutkan berdasarkan skor tertinggi
film_imdb_tinggi = film_imdb_tinggi.sort_values(
    by="IMDB Score",
    ascending=False
)

# Menampilkan hasil analisis
display(
    film_imdb_tinggi[
        ["Title", "Genre", "IMDB Score", "Language"]
    ]
)

print("Jumlah film dengan IMDb >= 8.0:",
      len(film_imdb_tinggi))

# Menampilkan film dengan skor tertinggi
if not film_imdb_tinggi.empty:
    print("Film dengan skor IMDb tertinggi:")
    display(
        film_imdb_tinggi[
            ["Title", "IMDB Score"]
        ].head(1)
    )
else:
    print("Tidak ada film dengan skor IMDb >= 8.0.")
```


### 4. Analisis Genre Film Netflix Original Tahun 2020

Analisis ini bertujuan untuk mengetahui genre yang paling banyak muncul dalam film Netflix Original yang dirilis pada tahun 2020. Dataset difilter berdasarkan tahun rilis pada kolom `Premiere`, kemudian jumlah film untuk setiap genre dihitung dan diurutkan dari yang terbanyak.

```python

# Memfilter film yang dirilis pada tahun 2020
film_2020 = df_clean[
    df_clean["Premiere"].dt.year == 2020
].copy()

# Menghitung jumlah film berdasarkan genre
jumlah_genre_2020 = (
    film_2020["Genre"]
    .value_counts()
    .reset_index()
)

jumlah_genre_2020.columns = ["Genre", "Jumlah Film"]

# Menampilkan 10 genre terbanyak
display(jumlah_genre_2020.head(10))

# Menampilkan genre terbanyak
if not jumlah_genre_2020.empty:
    print("Genre terbanyak pada tahun 2020:",
          jumlah_genre_2020.iloc[0]["Genre"])

    print("Jumlah film:",
          jumlah_genre_2020.iloc[0]["Jumlah Film"])
else:
    print("Tidak ditemukan film yang dirilis pada tahun 2020.")
```
