```python
import pandas as pd

df = pd.read_csv("../data/NetflixOriginals_UTF8.csv")
```

# Data Cleaning

Tahap Data Cleaning dilakukan untuk mempersiapkan dataset Netflix Original Films & IMDb Scores sebelum proses Exploratory Data Analysis (EDA). Berdasarkan hasil Data Understanding, dataset terdiri dari 584 baris dan 6 kolom, tanpa missing value maupun data duplikat. Oleh karena itu, proses cleaning difokuskan pada pemeriksaan dan penyesuaian format tanggal rilis pada kolom `Premiere` agar dapat digunakan dalam analisis berdasarkan tahun rilis film.

### 1. Menyiapkan Dataset

Dataset asli disalin terlebih dahulu agar tetap terjaga selama proses Data Cleaning. Seluruh proses perbaikan selanjutnya dilakukan pada salinan dataset.

```python
#Membuat salinan dataset
df_clean = df.copy()

#Menampilkan ukuran dataset
print("Ukuran dataset:", df_clean.shape)
```

### 2. Pemeriksaan Kolom Premiere

Kolom `Premiere` berisi informasi tanggal rilis film Netflix Original. Pemeriksaan dilakukan untuk mengetahui format tanggal asli dan mengidentifikasi nilai yang berpotensi menyebabkan kegagalan konversi ke tipe data tanggal.

```python
# Memeriksa contoh tanggal asli
display(df_clean[["Title", "Premiere"]].head(10))

#Memeriksa tipe data awal
print("Tipe data premiere:", df_clean["Premiere"].dtype)
```

### 3. Identifikasi Tanggal yang Gagal Dikonversi

Kolom `Premiere` dikonversi sementara ke format datetime untuk mengidentifikasi tanggal yang tidak sesuai dengan format yang dapat dibaca oleh Pandas. Nilai yang gagal dikonversi ditandai sebagai `NaT` dan diperiksa menggunakan data asli agar format tanggal sebelum perbaikan tetap dapat diketahui.

```python
# Mengonversi tanggal pada salinan sementara
premiere_datetime = pd.to_datetime(
    df_clean["Premiere"],
    errors="coerce"
)

# Membuat penanda tanggal yang gagal dikonversi
mask_tanggal_invalid = premiere_datetime.isna()

# Menampilkan tanggal asli yang bermasalah
tanggal_invalid = df_clean.loc[
    mask_tanggal_invalid, 
    ["Title", "Premiere"]
]

display(tanggal_invalid)

print("Jumlah tanggal yang gagal dikonversi:",
      mask_tanggal_invalid.sum())
```

### 4. Perbaikan Format Tanggal Rilis

Berdasarkan hasil pemeriksaan, ditemukan lima tanggal yang menggunakan tanda titik setelah angka tanggal. Format tersebut diperbaiki dengan mengganti tanda titik menjadi koma, kemudian kolom `Premiere` dikonversi ke tipe data datetime. Perbaikan dilakukan untuk menyeragamkan format tanggal tanpa menghapus data film.

```python
# Memperbaiki format tanggal pada dataset
df_clean["Premiere"] = (
    df_clean["Premiere"].
    str.replace(
        r"(\d{1,2})\.\s*(\d{4})",
        r"\1, \2",
        regex=True
    )
)

# Mengonversi tanggal menjadi datetime
df_clean["Premiere"] = pd.to_datetime(
    df_clean["Premiere"],
    errors="coerce"
)

# Memeriksa hasil perbaikan
print("Tipe data Premiere:",
      df_clean["Premiere"].dtype)

print("Tanggal gagal dikonversi:",
      df_clean["Premiere"].isna().sum())
```

### 5. Validasi Hasil Data Cleaning

Validasi dilakukan untuk memastikan dataset tetap memiliki ukuran yang sesuai setelah proses perbaikan tanggal. Pemeriksaan meliputi missing values, data duplikat, dan tipe data setiap kolom. Hasil validasi digunakan untuk memastikan dataset siap digunakan dalam tahap EDA.

```python
print("=== HASIL DATA CLEANING ===")

# Ukuran dataset
print("Ukuran dataset:", df_clean.shape)

# Missing values
print("\nMissing values setiap kolom:")
display(df_clean.isnull().sum())

# Data duplikat
print("\nJumlah data duplikat:",
      df_clean.duplicated().sum())

# Tipe data
print("\nTipe data setiap kolom:")
display(df_clean.dtypes)

# Menyimpan dataset hasil cleaning
df_clean.to_csv("../data/netflix_clean.csv", index=False)

print("Dataset hasil cleaning berhasil disimpan.")
```
