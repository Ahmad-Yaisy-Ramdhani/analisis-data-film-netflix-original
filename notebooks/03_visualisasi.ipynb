### 1. Visualisasi Bahasa Film Netflix Original

Visualisasi ini menggunakan diagram batang horizontal untuk menampilkan 10 bahasa yang paling banyak digunakan dalam film Netflix Original. Panjang batang menunjukkan jumlah film untuk setiap bahasa, sehingga bahasa dengan jumlah film terbanyak lebih mudah dibandingkan.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Membaca dataset hasil cleaning
df_clean = pd.read_csv(
    "../data/netflix_clean.csv",
    parse_dates=["Premiere"]
)

# Menghitung 10 bahasa terbanyak
bahasa_top10 = df_clean["Language"].value_counts().head(10)

# Membuat visualisasi
plt.figure(figsize=(10, 6))
bahasa_top10.sort_values().plot(kind="barh")

plt.title("10 Bahasa yang Paling Banyak Digunakan dalam Film Netflix Original")
plt.xlabel("Jumlah Film")
plt.ylabel("Bahasa")
plt.tight_layout()
plt.show()
```

### 2. Visualisasi Film dengan Skor IMDb 8.0 ke Atas

Visualisasi ini menampilkan 10 film dengan skor IMDb tertinggi dari film-film yang memperoleh skor minimal 8.0. Diagram batang horizontal digunakan agar judul film lebih mudah dibaca dan perbedaan skor IMDb dapat dibandingkan.

```python
# Memfilter film dengan skor IMDb minimal 8.0
film_imdb_tinggi = df_clean[
    df_clean["IMDB Score"] >= 8.0
].copy()

# Mengambil maksimal 10 film dengan skor tertinggi
film_top10 = film_imdb_tinggi.nlargest(
    10, "IMDB Score"
).sort_values("IMDB Score")

# Membuat visualisasi
plt.figure(figsize=(10, 7))
plt.barh(
    film_top10["Title"],
    film_top10["IMDB Score"]
)

plt.title("10 Film Netflix Original dengan Skor IMDb Tertinggi (Minimal 8.0)")
plt.xlabel("Skor IMDb")
plt.ylabel("Judul Film")
plt.xlim(8.0, 10.0)
plt.tight_layout()
plt.show()
```

### 3. Visualisasi Genre Film Netflix Original Tahun 2020

Visualisasi ini menampilkan 10 genre terbanyak dari film Netflix Original yang dirilis pada tahun 2020. Diagram batang horizontal digunakan untuk membandingkan jumlah film berdasarkan kategori genre dan mengidentifikasi genre yang paling dominan.

```python
# Memfilter film yang dirilis pada tahun 2020
film_2020 = df_clean[
    df_clean["Premiere"].dt.year == 2020
]

# Menghitung 10 genre terbanyak
genre_top10 = (
    film_2020["Genre"]
    .value_counts()
    .head(10)
)

# Membuat visualisasi
plt.figure(figsize=(10, 6))
genre_top10.sort_values().plot(kind="barh")

plt.title("10 Genre Film Netflix Original Terbanyak pada Tahun 2020")
plt.xlabel("Jumlah Film")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
```
