```python
import pandas as pd

df = pd.read_csv("../data/NetflixOriginals_UTF8.csv")
```

```python
print("=== UKURAN DATASET ===")
print("Jumlah baris dan kolom:", df.shape)

print("\n=== TIPE DATA ===")
print(df.dtypes)

print("\n=== MISSING VALUE ===")
print(df.isnull().sum())

print("\n=== DATA DUPLIKAT ===")
print(df.duplicated().sum())

print("\n=== STATISTIK DATA NUMERIK ===")
print(df.describe())

print("\n=== JUMLAH NILAI UNIK ===")
print(df.nunique())
```

```python
# Memeriksa nilai unik pada kolom teks
print("=== CONTOH NILAI UNIK ===")
print("Genre:")
print(df["Genre"].value_counts().head(10))

print("\nLanguage:")
print(df["Language"].value_counts().head(10))

# Memeriksa rentang skor IMDb
print("\n=== RENTANG SKOR IMDb ===")
print("Minimum:", df["IMDB Score"].min())
print("Maksimum:", df["IMDB Score"].max())

# Memeriksa durasi film
print("\n=== RENTANG RUNTIME ===")
print("Minimum:", df["Runtime"].min())
print("Maksimum:", df["Runtime"].max())

# Memeriksa tanggal tayang
print("\n=== CONTOH TANGGAL PREMIERE ===")
print(df["Premiere"].head(10))
```
